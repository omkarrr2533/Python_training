```python
days = [1,2,3,4,5]
a = [10,20,30,40,50]
b = [8,14,18,20,26]

plt.plot(days, a, color = "blue", marker="o", linestyle="--", label= "Python")
plt.plot(days, b, color = "green", marker="o", linestyle="--", label= "Java")

plt.title("weekly Scores")
plt.xlabel("day")
plt.ylabel("Score")
"""plt.legend() """
plt.grid(True)
plt.show()
```

```python
sub = ["Math", "Science", "English", "History"]
marks = [85, 78, 92, 70]
plt.barh(sub, marks, color = "black")
plt.bar(sub, marks, color = "green", width = 1)

plt.title("Marks by Subject")
plt.legend()
plt.ylabel("Marks")
plt.show()
```

```python
np.random.seed(42)
data = np.random.normal(70, 10, 200)

plt.hist(data, bins = 10, color = "teal", edgecolor= "black")
plt.title("distribution of marks")
plt.xlabel("Marks")
plt.ylabel("students")
plt.show()
```

```python
hours = [1,2,3,4,5,6,7,8]
score = [35,40,56,78,52,96,78,88]

plt.scatter(hours, score, color = "red")            
plt.title("study Hours vs Scores")
plt.xlabel("Hours")
plt.ylabel("Score")
plt.show()
```

```python
sizes = [40,30,20,10]
names = ["Python", "java", "SQL", "Other"]

plt.pie(sizes, labels = names, autopct = "%1.1f%%")
plt.title("class interest")
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

days = [2, 3, 8, 9, 6, 4, 7, 5]
temp = [25, 30, 26, 42, 35, 38, 29, 31]  

ax[0].plot(days, temp)
ax[0].set_title("Line")

subjects = ['Math', 'Science', 'English', 'History']
marks = [85, 90, 78, 88]

ax[1].bar(subjects, marks)
ax[1].set_title("Bar")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd 

df = pd.DataFrame({
    "name": ["Riya", "aman", "Neha", "karan", "Isha","Rohan"],
    "city": ["pune", "csn", "mumbai", "pune","kolhapur","mumbai"],
    "marks": [85,45,96,75,12,86],
    "attendance": [92,75,86,45,69,84]
})

df.plot(x = "name", y = "marks", kind= "bar", legend= False, title = "marks of students")
plt.ylabel("marks")
plt.show()
```

```python
df.groupby("city")["marks"].mean().plot(kind="bar", title = "Average Marks by City")
plt.ylabel("Average")
plt.show()
```

```python
df.groupby("city")
```

```python
import seaborn as sn
sn.set_theme(style="whitegrid")

sn.barplot(data=df, x="name", y = "marks")
plt.title("Marks(Seaborn)")
plt.show()
```

```python
sn.countplot(data = df, x = "city")
plt.title("Students per City")
plt.show()
```

```python
import pandas as pd
import matplotlib as plt
import seaborn as sn  # <-- Custom alias

if "df" not in globals():
    df = pd.DataFrame({
        "name": ["Riya", "aman", "Neha", "karan", "Isha", "Rohan"],
        "city": ["pune", "csn", "mumbai", "pune", "kolhapur", "mumbai"],
        "marks": [85, 45, 96, 75, 12, 86],
        "attendance": [92, 75, 86, 45, 69, 84]
    })

sn.scatterplot(data=df, x="attendance", y="marks", hue="city", s=120)
plt.grid(True)
plt.show()
```

```python
sn.histplot(data=df, x="marks", bins= 5,kde = True)
plt.title("Marks Distribution")
plt.show()
```

```python
sn.boxplot(data= df, x="city", y = "marks")
import matplotlib.pyplot as plt
plt.title("marks by City")
plt.show()
```

```python
sn.heatmap(df[["marks", "attendance"]], annot=True, cmap="YlGnBu", fmt="d")
plt.title("Marks by City")
plt.show() 
```
