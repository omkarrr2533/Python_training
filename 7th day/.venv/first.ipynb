```python
%pip install matplotlib

import matplotlib.pyplot as plt
import seaborn as sea
import pandas as pd
import numpy as np 
```

```python
days = [1,2,3,4,5]
a = [10,20,30,40,50]
b = [8,14,18,20,26]

plt.plot(days, a, color = "blue", marker="o", linestyle="--", label= "Python")
plt.plot(days, b, color = "green", marker="o", linestyle="--", label= "Java")

plt.title("weekly Scores")
plt.xlabel("day")
plt.ylabel("Score")
"""plt.legend() """
plt.grid(True)
plt.show()
```

```python
sub = ["Math", "Science", "English", "History"]
marks = [85, 78, 92, 70]
plt.barh(sub, marks, color = "black")
plt.bar(sub, marks, color = "green", width = 1)

plt.title("Marks by Subject")
plt.legend()
plt.ylabel("Marks")
plt.show()
```

```python
np.random.seed(42)
data = np.random.normal(70, 10, 200)

plt.hist(data, bins = 10, color = "teal", edgecolor= "black")
plt.title("distribution of marks")
plt.xlabel("Marks")
plt.ylabel("students")
plt.show()
```

```python
hours = [1,2,3,4,5,6,7,8]
score = [35,40,56,78,52,96,78,88]

plt.scatter(hours, score, color = "red")            
plt.title("study Hours vs Scores")
plt.xlabel("Hours")
plt.ylabel("Score")
plt.show()
```

```python
sizes = [40,30,20,10]
names = ["Python", "java", "SQL", "Other"]

plt.pie(sizes, labels = names, autopct = "%1.1f%%")
plt.title("class interest")
plt.show()
```
