```python
pop = [10,15,20,5,30]
cities = ["Pune", "mumbai", "CSN", "Nagar","Nagpur"]

plt.bar(cities, pop, color="green")
plt.title("City vs Population Chart")
plt.xlabel("cities")
plt.ylabel("population")
plt.show()
```

```python
import matplotlib.pyplot as plt

activities = ["Sleep", "College/Work", "Study", "Travel", "Meals", "Other"]
hours = [8, 7, 4, 2, 2, 1]

plt.pie(hours, labels=activities, autopct="%1.1f%%", startangle=90)
plt.title("How I Spend 24 Hours a Day")
plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 2, 6, 7, 8, 5]
marks = [40, 50, 55, 65, 75, 48, 80, 85, 95, 70]

plt.scatter(study_hours, marks, color="blue")
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
python = [60, 70, 75, 85, 90]
java = [55, 65, 80, 75, 88]

plt.plot(days, python, marker="o", label="Python")
plt.plot(days, java, marker="s", label="Java")

plt.title("Python vs Java Scores")
plt.xlabel("Day")
plt.ylabel("Score")
plt.legend()
plt.grid(True)
plt.show()
```
